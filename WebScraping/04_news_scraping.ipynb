```python
import os
import requests
from bs4 import BeautifulSoup

def download_one_episode(title, no, url):
    h = {'User-Agent': 'Mozilla/5.0'}
    r = requests.get(url, headers=h)
    r.encoding = 'utf-8'
    s = BeautifulSoup(r.text, 'html.parser')
    i = s.select('div.wt_viewer img')
    p = os.path.join('img', title, str(no))
    os.makedirs(p, exist_ok=True)
    for x, v in enumerate(i):
        u = v['src']
        e = os.path.splitext(u)[1].split('?')[0] or '.jpg'
        f = os.path.join(p, f'{x+1:03}{e}')
        try:
            with open(f, 'wb') as out:
                out.write(requests.get(u, headers=h).content)
        except:
            pass
```

```python
download_one_episode('낢이사는이야기', 48, 'https://comic.naver.com/webtoon/detail?titleId=833255&no=49&week=tue')
```
