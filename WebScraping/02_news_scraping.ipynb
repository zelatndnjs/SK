```python
import requests
from bs4 import BeautifulSoup

section_dict = {
    '기후/환경': 'climate', '사회': 'society', '경제': 'economy', '정치': 'politics',
    '국제': 'world', '문화': 'culture', '생활': 'life', 'IT/과학': 'tech', '인물': 'people'
}

def print_news(section_name):
    section_code = section_dict.get(section_name)
    if not section_code:
        print(f"'{section_name}' 섹션 없음")
        return

    url = f'https://news.daum.net/{section_code}'
    print(f"======> {url} {section_name} 뉴스 <======\n")
    
    response = requests.get(url)
    response.encoding = 'utf-8'
    soup = BeautifulSoup(response.text, 'html.parser')

    news_list = soup.select('ul.list_newsheadline2 li')

    for li in news_list:
        a_tag = li.find('a')
        link = a_tag['href']
        title_tag = li.select_one('div.cont_thumb strong.tit_txt')
        title = title_tag.text.strip() if title_tag else '제목 없음'
        print(f"{title}\n{link}\n")

print_news('경제')
print_news('사회')
```
