```python
import requests as r
from bs4 import BeautifulSoup as b
from urllib.parse import urljoin
from IPython.display import Image, display
import re

sec = {
    '최신뉴스': 'n0100',
    '정치': 'n0101',
    '경제': 'n0102',
    '사회': 'n0103',
    '세계': 'n0104',
    'IT/과학': 'n0105'
}

def g(u):
 if not u: return None
 s = u.get('src','')
 if s.startswith('//') and '///' not in s: return 'https:' + s
 o = u.get('onerror','')
 m = re.search(r"ImgError\(this,'(//[^']+)'", o)
 if m: return 'https:' + m.group(1)
 return None

def go():
 for k,v in sec.items():
  print(f'\n[{k}] https://news.nate.com/recent?mid={v}\n')
  h = r.get(f'https://news.nate.com/recent?mid={v}')
  h.encoding = 'euc-kr'
  s = b(h.text, 'html.parser')
  t = s.select('h2.tit')
  for i,x in enumerate(t[:2]):
   print(f'{i+1}.', x.text.strip())
   a = x.find_parent('a')
   l = urljoin(h.url, a['href']) if a and a.has_attr('href') else ''
   print(l)
   im = g(x.find_previous('img'))
   if im: display(Image(url=im))
   else: print('img x')
   print('-'*40)

go()
```
